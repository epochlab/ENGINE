```python
from dataclasses import replace
import matplotlib.pyplot as plt

from pathtracer import Renderer, display_encode
```

```python
renderer = Renderer("scenes/cornell.json")
```

```python
cam = replace(renderer.default_camera, position=(0.0, 0.0, 6.0))
frame = renderer.render(camera=cam, aovs=("beauty", "depth", "normal"), width=128, height=64, samples=16, seed=1, show_sky=False)
```

```python
x = frame["beauty"] #float 32
```

```python
plt.imshow(display_encode(x)) # Scene-referred linear to display-referred sRGB
plt.xticks([]), plt.yticks([])
plt.tight_layout()
print(f"Min: {x.min()}, Max: {x.max()}")
```
